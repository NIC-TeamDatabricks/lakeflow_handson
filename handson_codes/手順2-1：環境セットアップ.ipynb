# Lakeflowパイプライン トレーニング環境セットアップ

このノートブックは、Lakeflow Spark Declarative Pipelinesのトレーニング環境をセットアップします。

## アーキテクチャ
- **Bronze層**: Auto Loaderで生データを取り込み
- **Silver層**: データクレンジング、品質ルール適用、Merge差分更新
- **Gold層**: ビジネス用データマート

## セットアップ内容
1. トレーニング用カタログの作成
2. 各層のスキーマ作成（bronze, silver, gold）
3. ソースデータ用Volume作成
4. 環境の確認

```python
# トレーニング環境の設定

# カタログとスキーマの命名
catalog_name = f"lakeflow_training"
bronze_schema = "bronze_schema"
silver_schema = "silver_schema"
gold_schema = "gold_schema"
volume_name = "source_data"

print(f"カタログ名: {catalog_name}")
print(f"Bronze スキーマ: {bronze_schema}")
print(f"Silver スキーマ: {silver_schema}")
print(f"Gold スキーマ: {gold_schema}")
print(f"Volume名: {volume_name}")
```

```python
# トレーニング用カタログの作成
spark.sql(f"""
  CREATE CATALOG IF NOT EXISTS {catalog_name}
  COMMENT 'Lakeflow Pipeline トレーニング用カタログ'
""")
print(f"カタログ '{catalog_name}' を作成しました")
```

```python
# Bronze層: 生データを格納
spark.sql(f"""
  CREATE SCHEMA IF NOT EXISTS {catalog_name}.{bronze_schema}
  COMMENT 'Bronze層 - Auto Loaderで取り込んだ生データを格納'
""")
print(f"スキーマ '{catalog_name}.{bronze_schema}' を作成しました")
```

```python
# Silver層: クレンジング済みデータを格納
spark.sql(f"""
  CREATE SCHEMA IF NOT EXISTS {catalog_name}.{silver_schema}
  COMMENT 'Silver層 - クレンジング済み、品質ルール適用済みデータ'
""")
print(f"スキーマ '{catalog_name}.{silver_schema}' を作成しました")
```

```python
# Gold層: ビジネス用データマートを格納
spark.sql(f"""
  CREATE SCHEMA IF NOT EXISTS {catalog_name}.{gold_schema}
  COMMENT 'Gold層 - ビジネス分析用データマート'
""")
print(f"スキーマ '{catalog_name}.{gold_schema}' を作成しました")
```

```python
# ソースデータを配置するVolumeを作成
# Bronze層スキーマ内に作成（データ取り込み元として使用）
spark.sql(f"""
  CREATE VOLUME IF NOT EXISTS {catalog_name}.{bronze_schema}.{volume_name}
  COMMENT 'Auto Loader用のソースデータを格納するVolume'
""")
print(f"Volume '{catalog_name}.{bronze_schema}.{volume_name}' を作成しました")
```
