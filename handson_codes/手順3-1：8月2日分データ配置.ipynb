# 8月2日分テレマティクスデータの配置

このノートブックは、8月2日分のテレマティクスデータをVolumeに配置します。

## 動作
1. ソースファイル：`demo_data/telematics_data_20260802.csv`
2. コピー先：`/Volumes/lakeflow_training/bronze_schema/source_data/`
3. **ファイルトリガーがアクティブの場合、パイプラインが自動起動します**

## 確認事項
- ファイルトリガーがアクティブに設定されているか
- パイプラインが起動するか（ワークフロー画面で確認）

```python
# 8月2日分のテレマティクスデータをVolumeにコピー

# 現在のユーザー名を取得
current_user = spark.sql("SELECT current_user()").collect()[0][0]

# ソースファイルパス
source_csv = f"/Workspace/Users/{current_user}/lakeflow_handson/handson_codes/demo_data/telematics_data_20260802.csv"

# コピー先Volume（Lakeflowパイプラインの取り込み元）
volume_path = "/Volumes/lakeflow_training/bronze_schema/source_data"
target_csv = f"{volume_path}/telematics_data_20260802.csv"

print(f"ソース: {source_csv}")
print(f"コピー先: {target_csv}")
print()

# ファイルをVolumeにコピー
try:
    dbutils.fs.cp(source_csv, target_csv, recurse=False)
    print("✓ ファイルのコピーが完了しました")
    print()
    print("⚠️ ファイルトリガーがアクティブの場合、パイプラインが自動的に起動します。")
    print("👉 ワークフロー画面でパイプラインの実行状況を確認してください。")
    
    # Volume内のファイル一覧を確認
    print(f"\n【{volume_path} 内のファイル一覧】")
    files = dbutils.fs.ls(volume_path)
    for file in files:
        print(f"  - {file.name} ({file.size:,} bytes)")
        
except Exception as e:
    print(f"エラー: {e}")
```
